```python
import scipy.stats as st
import numpy as np
import matplotlib.pyplot as plt
```

zadanie 4

```python
theta = 5
tab_mm, tab_mle, tab_i = [], [], []
for i in range(10, 1000):#[2**i for i in range(21)]:
    tab = st.expon.rvs(loc=theta, size=i)
    tab_mm.append(np.mean(tab)-1)
    tab_mle.append(min(tab))
    tab_i.append(i)

plt.scatter(tab_i, tab_mm, color="blue", label="MM", s=3)
plt.scatter(tab_i, tab_mle, color="green", label="MLE", s=3)
plt.plot([0, max(tab_i)], [theta,theta], linestyle="--", color="red", alpha=0.7, label="$\\theta$")
plt.ylabel("wartośc estymatora i parametru")
plt.xlabel("rozmiar próby")
plt.legend()
#plt.loglog()
plt.show()
```

zadanie 5

```python
def z9_rvs(a, n=1):
    tab = st.uniform.rvs(size=n)
    return tab**(1/(a+1))
```

```python
alfa = 2
tab_mm, tab_mle, tab_i = [], [], []
for i in range(10, 1000):#[2**i for i in range(21)]:
    tab = z9_rvs(alfa, i)
    tab_mm.append(-1-np.mean(tab)/(np.mean(tab)-1))
    tab_mle.append(-1-len(tab)/sum(np.log(tab)))
    tab_i.append(i)

plt.scatter(tab_i, tab_mm, color="blue", label="MM", s=3)
plt.scatter(tab_i, tab_mle, color="green", label="MLE", s=3)
plt.plot([0, max(tab_i)], [alfa, alfa], linestyle="--", color="red", alpha=0.7, label="$\\theta$")
plt.xlabel("wartośc estymatora i parametru")
plt.ylabel("rozmiar próby")
plt.legend()
#plt.loglog()
plt.show()
```

```python
k, theta = 5,2
fig, ax = plt.subplots(nrows=3, ncols=2, constrained_layout=True, figsize=(6,7))
tab_s = [100, 1000, 10000]

for i in range(3):
    tab_k, tab_theta, tab_i = [], [], []
    for j in range(10, 1000):#[2**i for i in range(21)]:
        tab = st.gamma.rvs(a=k, scale=theta, size=tab_s[i])
        tab_theta.append(np.var(tab)/np.mean(tab))
        tab_k.append(np.mean(tab)/tab_theta[-1])
        tab_i.append(j)
    ax[i][0].scatter(tab_i, tab_k, color="blue", label="estymator k", s=3)
    ax[i][0].plot([0, max(tab_i)], [k, k], linestyle="--", color="red", alpha=0.7, label="k")
    ax[i][1].scatter(tab_i, tab_theta, color="green", label="estymator $\\theta$", s=3)
    ax[i][1].plot([0, max(tab_i)], [theta, theta], linestyle="--", color="red", alpha=0.7, label="$\\theta$")
    ax[i][0].set_title(f"średnia k: {np.mean(tab_k):.3f}")
    ax[i][1].set_title(f"średnia $\\theta$: {np.mean(tab_theta):.3f}")

ax[0][0].legend()
fig.suptitle("N=100,1000,10000")
plt.show()
```
