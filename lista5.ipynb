Wykorzystując symulacje Monte Carlo sprawdź poprawność estymatora.

Wyniki dla metody momentów porównaj z tymi uzyskanymi dla metody największej wiarogodności.

```python
def est1(X):
    return np.mean(X) - 1

def est2(X):
    return np.min(X)

def mse(X, EX):
    return np.mean((X - EX)**2)
```

```python

beta = 6
X = sc.expon.rvs(size=(100, 1000), loc=beta)
E1 = [est1(x) for x in X]
E2 = [est2(x) for x in X]

plt.scatter(np.arange(0, 100), E1, label='MM')

plt.scatter(np.arange(0, 100), E2, label='MLE')

print(np.mean(E1), np.mean(E2))
print(np.var(E1), np.var(E2))
plt.legend()
plt.show()
```
